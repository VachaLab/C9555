#### 3.2.1. Working with a terminal

<div class="code alert alert-block alert-info">

<div class="code alert alert-block alert-info">
<code>ls -l</code>
</div>

This will display information on each file or directory present in the current location, including:
* permissions on the given file (`-`) or directory (`d`) to be read (`r`), written (`w`), or executed (`x`) (e.g. `drwx------`, `-rw-rw-r--`, <i>etc.</i>),
* number of _hard links_, which is the number of files and directories
* owner name of the file or directory,
* owner group,
* file size in bytes (by default),
* time of last modification,
* file or directory name

#### 3.2.3. Creating symbolic links to files
Since some of the files you will generate through your simulations will be big, copying them might be inefficient. However, you can refer to these files by creating a (soft or symbolic) link, or symlink, that points to their path. For instance:

<div class="code alert alert-block alert-info">
<p><code>touch myBigFile</code></p>
<p><code>mkdir anotherDirectory</code></p>
<p><code>cd anotherDirectory</code></p>
<p><code>ln -s ../myBigFile myLink</code></p>
<p><code>ls -la</code></p>
</div>

<div class="alert alert-block alert-danger">
    <b>Important:</b> By default, <code>ln</code> creates hard links. The difference between hard link and soft links is that hard links share the same <a href="https://en.wikipedia.org/wiki/Inode">inode</a> while soft links don't:
<ul>
    <li>Hard links have thus many limitations: they cannot be used for directories, for files across different filesystems or partitions,</li>
    <li>However, if you create a hard link and remove the original file, you will only remove the initial hard link and thus you will still be able to access the file using the second hard link. That is not the case with symlinks. If you remove or move the initial file, all the soft links that pointed to it become unusable.</li>
</ul>
</div>

#### 3.2.4. Viewing and editing files
There are many programs and ways to edit files depending on the situation. 

##### 3.2.4.1. **Gedit, a visual WYSIWYG (what you see is what you get) text editor** 

The first option is to use a text editor with a graphical user interface (GUI). `gedit` is one of them. You can open gedit like you would open any application on Debian-based OS, by accessing the search bar after pressing the Windows key (or Super key), and typing the name of the application. The default path for opening and saving files will be your home directory.

You can also run gedit from a terminal and directly edit the file you want:
<div class="code alert alert-block alert-info">
<code>nohup gedit myNewFile.dat &</code>
</div>

Let's break down this previous command line:
* **`nohup`** is a command that means "no hang up". This command allows to execute another command, _e.g._ `gedit`, and to avoid terminating it if the user (you) logs out (closes) the terminal where the command was executed.
* **`&`** is a symbol that is added at the end of a command executed in a terminal. It asks to run the terminal in background, so that one can continue using the terminal for other tasks.

<div class="alert alert-block alert-success">
<b>Note:</b>
    If you forget to use the ampersand symbol (<code>&</code>), you can place a task in the background by using the shortcut <code>CTRL + z</code> which stops the process executed in foreground. You can then type the command <code>bg</code> (background) which places that stopped process in background.
</div>

Gedit is a simple editor, but many plugins exist that allow to make it a decent IDE for small projects. For instance, you can install plugins to obtain a "better" Python console, one to check Python scripts for errors, to swith between tabs with a shortcut, to fold/unfold code blocks, to automatically indent the code blocks, to interprete, to split the window vertically between different tabs, to check the spelling, to use auto-completion on code lines, to use versioning, or collaboration tools, _etc._

##### 3.2.4.2. **vi and vim**

If Gedit might be handy, there are different circumstances when you want to use an editor that is embedded in the terminal and that does not need to be ran in a separate interface. First, vim (the fork from vi distributed by default on many Debian-based distributions) is able to process large files faster than gedit. Second, in case you are accessing a distant file, opening a graphic application, if that is even possible, will use a substantial bandwidth while using vi will be much lighter and (almost) immediate.

<div class="alert alert-block alert-success">
<b>Note:</b>
    Alternative to vi/vim exist, such as Emacs or (GNU) nano. Choose wisely: https://en.wikipedia.org/wiki/Editor_war!
</div>

It is far beyond the scope of this practical to teach you any advanced use of vim, as this software presents many functions. However, you will learn how to open, edit, save and/or quit a file. For further resources, look at the links provided [above](#linux_environment_advanced_resources).

To open vim, type the command `vi` in a terminal. Vim functions with different modes which have different uses. The following table sums up these mods, their use, and how to access them:

| Mode | How to access it? | What to use it for?  |
| --- | --- | --- |
| Normal mode | Press `Esc` | Default mode, used to move the cursor in the file, some actions like copy, paste, delete, cancel past action, exit |
| Visual modes | Press `v` (visual), `CTRL + v` (block visual), or `SHIFT + v` (line visual) from normal mode | Visual selection of a portion of the document to copy, cut, or delete |
| Command mode | Press `:` from normal mode | save, exit, search for some pattern, substitute text, set options, record macros |
| Insert mode | Press `i`, `a`, or `o` or uppercase variants from normal mode | Allows to insert text |
| Replace mode | Press `r` uppercase variants from normal mode | Allows to replace some text |

In our case, we will use a very small amount of commands in a few of these modes to illustrate vim functioning. First, to insert some text, enter the Insert mode by pressing `i` (you can try the other options from the table above to enter this mode and see where it places the cursor relatively to using `i`).

As you have already noticed, vi/vim might not be the most user-friendly software to learn. To better illustrate that, here is a table which shows you how to exit vim depending on your situation:
| Situation | What to do? |
| --- | --- |
| Exit without saving an unedited file | Enter command mode and type `q`, validate by pressing `ENTER` |
| Exit without saving an edited file | Enter command mode and type `q!`, validate by pressing `ENTER` |
| Exit with saving, if the file is already named | Enter command mode and type `wq` (which is the combination of the commands save and exit), validate by pressing `ENTER` |
| Exit with saving, if the file is named (alternative 1) | Enter command mode and type `x`, validate by pressing `ENTER` |
| Exit with saving, if the file is named (alternative 2) | In normal mode and press `ZZ` |
| Exit with saving, if the file is not named | Enter command mode and type `x myFileName` (which is the combination of the commands save and exit), validate by pressing `ENTER` |

However, vim comes in handy rather quickly, as it allows to efficiently modify any file, once you cross the first steep learning slope. Here are a few examples of uses of vim which could teach you the basics in using this editor:
* In normal mode, `26yl` will copy (`y`) the next 26 characters (`l`); `3yw`, the next 3 words (`w`); and `yy`, only 1 line (the current one).
* The action `d` can be used instead of `y` to delete or cut instead of copying. 
* In normal mode, pressing `p` or `P` will paste the last copied or cut portion resp. before or after the cursor.
* The visual modes permit to select visually the portion to copy or cut.
* To search a pattern in a given file, you can use, in the command mode `/ pattern`. Then, pressing `n` or `N` in the normal mode moves the cursor to the next or previous occurence of the search.
* In normal mode, pressing `gg` makes you go to the first line of the file, pressing `G` to the last one.
* To go to the N-th line, use the command mode and indicate the line index. 

All of vim options can be easily overwhelming and take some time to get used, but vim remains a very often used and efficient text editor. 

##### 3.2.4.3. **More or less cat, head and tail**

There are a few more commands that allow you to view or edit files. These commands are often used for processing data, preparing simulations inputs, or monitor their outputs: 
* `less` is a simple viewer that allows to process large files, by loading it page-by-page
* `cat` is a a command that prints file(s) to the standard output, which means that the files processed by cat are consecutively written in the terminal were the command is invoked.
* `head` and `tail` are commands used to print only the beginning or the end of a file, by providing, for instance the number of lines that should be printed with the option `-n`.

##### 3.2.4.4. **Streams and pipes**

These past commands, and many others, can often be combined. For that purpose, Bash proposes a specific syntax to deal with data in different cases:
* **Redirections** of data are made by **stream redirections** (`<`, `>`, `>>`, `2>`, `2>>`, `2>&1`, `2>>&1`). You can use redirections in different ways:
    1. to provide some data to a command, _i.e._ the command `wc -l fileName` will return the number of lines in the file called fileName. You can also provide this input using using input stream redirection: `wc < fileName -l`. This syntax is often used when one wants to read some file line by line inside a loop.
    2. to redirect the output of the data from the standard output (_i.e._ the terminal where the command is executed) to some file, _e.g._ `ls -latrh > fileName`. If one wants to append a file, use `>>` instead. The other stream redirections allow to redirect not only the normal output but also the potential errors, either to a separate direction, or to the same one.
* **Pipes** are used to create chained commands. It will directly redirect the output of a first command to the input of a second one. For instance: `ls -latrh | grep ".py"` will list all files for a directory and search for all lines in this output with the pattern ".py".

### 4.3. Working with trajectories

The time-evolving coordinates of a system are called a trajectory. They are most commonly obtained in simulations of molecular systems, but can also be generated by other means and for different purposes. Upon loading a trajectory, VMD can animate movies of system evolving with time and analyze various structural features throughout the trajectory. This section will introduce the basics of working with trajectory data in VMD. 

#### 4.3.1. Loading trajectories

Trajectory files are typically large binary files that contain the time varying atomic coordinates for the system. Each set of coordinates corresponds to one frame in time. Examples of a trajectory files are DCD and XTC files. Trajectory files usually do not contain information structural information as found in protein structure files (PSF). Therefore, we must first load the structure file, and then add the trajectory data to the same molecule, so that VMD has access to both the structure and trajectory information.

1. Start a new VMD session. In the VMD Main window, select _File_ $\rightarrow$ _New Molecule..._. The Molecule File Browser window should appear on your screen. 
2. Use the _Browse..._ button to find the file `ubiquitin.psf` in the directory `week1`. When you select the file, you will be back in the Molecule File Browser window. Press the _Load_ button to load the molecule. 
3. In the Molecule File Browser window, make sure that `ubiquitin.psf` is selected in the _Load files for_: pull-down menu on the top, and click on the _Browse_ button. Browse for `pulling.dcd`. Note the options available in the Molecule File Browser window: one can load trajectories starting and finishing at chosen frames, and adjust the stride between the loaded frames. Leave the default settings so that the whole trajectory is loaded. 
4. Click on the _Load_ button in the Molecule File Browser window. You will be able to see the frames as they are loaded into the molecule. After the trajectory finishes loading, you will be looking at the last frame of your trajectory. To go to the beginning, use the animation tools at the lower part of the VMD Main menu (see Fig. 9). You can close the Molecule File Browser window.

#### 4.4.2. Example of built-in analysis tool: the RMSD Trajectory Tool

The built-in analysis tools in VMD are available under the menu item _Extensions_ $\rightarrow$ _Analysis_. Each of these tools features a GUI window that allows one to enter parameters and customize the quantities analyzed. In addition, all tools can be invoked in a scripting mode, using the TkConsole window. We will learn how to work with one of the most frequently used tools, the RMSD Trajectory Tool.

The Root Mean Squared Deviation (RMSD) is defined as:

\begin{equation}
RMSD = \sqrt{ \frac{ \sum_{i=1}^{N_{atoms}} || r_i(t_1) - r_i(t_2) ||^2 }{N_{atoms}}},
\end{equation}

where $r_i(t)$ is the position of atom i at time t.

In this example, we will analyze the RMSD for two trajectories for the same system, ubiquitin.psf. One of them is the already familiar pulling trajectory, `pulling.dcd`, and the other is the trajectory of a simulation in which no force was applied to the protein, `equilibration.dcd`.

1. Start a new VMD session. Load the ubiquitin equilibration trajectory into VMD (using the files ubiquitin.psf and equilibration.dcd). 

2. Choose _Extensions_ $\rightarrow$ _Analysis_ $\rightarrow$ _RMSD Trajectory Tool_ in the VMD Main window (Fig. 13). The RMSD Trajectory Tool window will show up. 

3. In the RMSD Trajectory Tool window, you can see many customizations can be made. For the default values, the molecule to be analyzed is `ubiquitin.psf` (the only one loaded). The selection for which RMSD will be computed is all of the `protein` atoms, excluding hydrogens (since the `noh` checkbox is on). The RMSD will be calculated for each frame (time $t_{1}$ in Eq. 1) with the reference to frame 0 (time $t_{2}$ in Eq. 1). Make sure the _Plot checkbox_ is selected.

#### 5.11.2. Recursive functions

In Python, it is possible to define recursive functions, that are functions that call themselves to solve a problem by breaking it down into smaller, similar sub-problems.

A recursive function consists of two parts:
1. Base Case: A condition that defines when the recursion should stop. It's the simplest case that can be solved directly without further recursion.
2. Recursive Case: The part of the function where it calls itself with a modified input to solve a smaller instance of the same problem.

Common examples include the computation of the factorial of a non-negative integer $n!$ or generating the Fibonacci sequence.

````python
def factorial(n):
    # Base case
    if n == 0:
        return 1
    
    # Recursive case
    else:
        return n * factorial(n - 1)
    
def fibonacci(n):
    # Base cases
    if n == 0:
        return 0
    elif n == 1:
        return 1
    
    # Recursive case
    else:
        return fibonacci(n - 1) + fibonacci(n - 2)
    
print(f"The value of 5! is {factorial(5)}.")

print("Here follows the first 6 values of the Fibonnaci sequence:")
for n in range(6):
    print(f"F_{n}={fibonacci(n)}")
````

#### 5.12.1. Reading text files

To read a text file in Python, you can use the `open()` function, which returns a file object. You can specify the file path and the mode (e.g., `r` for reading, `w` for writing, `a` for appending, and more).

```python
with open('example.txt', 'r') as file:

print(content)
```

#### 5.12.2. Writing text files

To write to a text file, use the `w` mode to overwrite a file, or write a new file, or the `a` mode to append some file with the `open()` function.

```python
# Writing to a text file
with open('output.txt', 'w') as file:
    file.write('Hello, world!')
```

#### 5.12.3. Handling column separators

Suppose you have a file with data separated by spaces, tabs, or an unknown number of spaces. You can split the lines into columns using Python's `split()` method. For instance, to read a space-separated file:

```python
with open('space_separated.txt', 'r') as file:
    for line in file:

        # Process the columns
```

For tab-separated files, replace `split()` with `split('\t')`.

````python
import requests

# Function to download a FASTA file based on the PDB identifier
def download_fasta(pdb_identifier):
    url = f"https://www.rcsb.org/fasta/entry/{pdb_identifier}"
    
    try:
        response = requests.get(url)
        response.raise_for_status()  # Raise an exception for HTTP errors

        if response.status_code == 200:
            content = response.text
            if "No valid PDB IDs were submitted." in content:
                print("No valid PDB IDs were submitted.")
                return None
            return content
        else:
            print("Failed to download the FASTA file.")
            return None

    except requests.exceptions.RequestException as e:
        print(f"Error: {e}")
        return None

# Function to find repeating motifs in a sequence
def find_repeating_motifs(sequence):
    motifs = []
    seen_motifs = set()  # Use a set to track unique motifs

    for length in range(3, len(sequence) + 1):
        for start in range(len(sequence) - length + 1):
            motif = sequence[start:start + length]
            repetitions = sequence.count(motif)

            if repetitions >= 3 and motif not in seen_motifs:
                motifs.append((motif, repetitions))
                seen_motifs.add(motif)

    # Sort motifs by length (descending) and then by repetitions (descending)
    motifs.sort(key=lambda x: (-len(x[0]), -x[1]))
    return motifs

def main():
    # Main menu loop
    protein_data = {}

    while True:
        print("\nMain Menu:")
        print("1. Analyze a new FASTA file")
        print("2. Retrieve previous analysis")
        print("3. Quit")

        choice = input("Enter your choice (1/2/3): ")

        if choice == "1":
            pdb_identifier = input("Enter PDB identifier: ")
            fasta_content = download_fasta(pdb_identifier)

            if fasta_content:
                proteins = fasta_content.strip().split(">")[1:]
                protein_info = {}

                for protein in proteins:
                    lines = protein.split("\n")
                    header = lines[0]
                    sequence = "".join(lines[1:])
                    protein_info[header] = sequence

                protein_data[pdb_identifier] = protein_info

                motifs_info = {}
                for header, sequence in protein_info.items():
                    motifs = find_repeating_motifs(sequence)
                    motifs_info[header] = motifs

                print("\nMotifs Analysis:")
                for header, motifs in motifs_info.items():
                    print(f"Protein: {header}")
                    if motifs:
                        for motif_info in motifs:
                            motif, repetitions = motif_info
                            if repetitions == 1:
                                print(f"Motif: {motif}, Repeat: {repetitions}")
                            else:
                                print(f"Motif: {motif}, Repeats: {repetitions}")
                    else:
                        print("No motifs found in this protein sequence.")

            else:
                print("Failed to analyze the FASTA file.")

        elif choice == "2":
            if not protein_data:
                print("No previous analysis results available.")
            else:
                print("\nPrevious Analysis Results:")
                for pdb_id, protein_info in protein_data.items():
                    print(f"PDB Identifier: {pdb_id}")
                    for header, sequence in protein_info.items():
                        motifs = find_repeating_motifs(sequence)
                        print(f"Protein: {header}")
                        if motifs:
                            print("Motifs Analysis:")
                            for motif_info in motifs:
                                motif, repetitions = motif_info
                                if repetitions == 1:
                                    print(f"Motif: {motif}, Repeat: {repetitions}")
                                else:
                                    print(f"Motif: {motif}, Repeats: {repetitions}")
                        else:
                            print("No motifs found in this protein sequence.")

        elif choice == "3":
            print("Exiting the program.")
            break

        else:
            print("Invalid choice. Please select 1, 2, or 3.")
            
if __name__ == "__main__":
    main()
````
